```python
class HashTable:
    def __init__(self, size=10):
        self.size = size
        self.table = [None for _ in range(size)]

    def hashKey(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        bucket = self.hashKey(key)
        print(f"Hashing key: {key} to bucket: {bucket}")
        if type(self.table[bucket]) == type([]):
            for i, (k, v) in enumerate(self.table[bucket]):
                if k == key:
                    self.table[bucket].append((key, value))
                    return True
        elif type(self.table[bucket])== type(()):
            if self.table[bucket][0] == key:
                print("Collision detected, updating entry to Chaining Method")
                existing_entry = self.table[bucket]
                self.table[bucket] = [existing_entry]
                self.table[bucket].append((key, value))
                return True
        elif self.table[bucket] is None:
            self.table[bucket] = (key, value)
            return True
        return False

    def get(self, key):
        bucket = self.hashKey(key)
        if type(self.table[bucket]) == type([]):
            returned_entries = ''
            for i, (k, v) in enumerate(self.table[bucket]):
                if k == key:
                    returned_entries += f"{str(v)}" + (", " if i < len(self.table[bucket]) - 1 else "")
            return returned_entries
        elif type(self.table[bucket])== type(()):
            if self.table[bucket][0] == key:
                return self.table[bucket][1]
        print(f"Key: {key} not found in bucket: {bucket}")
        return None
    
    def remove(self, key):
        bucket = self.hashKey(key)
        if type(self.table[bucket]) == type([]):
            for i, (k, v) in enumerate(self.table[bucket]):
                if k == key:
                    self.table[bucket] = None
                    return True
        elif type(self.table[bucket])== type(()):
            if self.table[bucket][0] == key:
                self.table[bucket] = None
                return True
        print(f"Key: {key} not found in bucket: {bucket}")
        return False
```

```python
ht = HashTable()
ht.insert("name", "Alice")
ht.insert("interests", "Cats")
ht.insert("name", "Bob")
ht.insert("city", "New York")
ht.insert("interests", "Orange Cats")
ht.insert("interests", "Cars")
ht.insert("city", "Raleigh")
ht.insert("images", "Custom Computers")
print("---Hash Table---")
for index in ht.table:
    print(f"{index}")
```
