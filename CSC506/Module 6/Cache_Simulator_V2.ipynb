```python

class LRU_Cache:
    def __init__(self, cache_size=10):
        self.cache = OrderedDict()
        self.cache_hits = 0
        self.cache_misses = 0
        self.cache_size = cache_size

    def get(self, key):
        if self.cache.get(key, False) != False:
            self.cache_hits += 1
            self.cache.move_to_end(key)  # Move the accessed item to the end
            return self.cache[key]
        else:
            self.cache_misses += 1
            data_from_db = global_database.get(key, False)
            if data_from_db != False:
                if len(self.cache) >= self.cache_size:
                    self.evict()
                self.cache[key] = data_from_db
                return data_from_db
            
    def evict(self):
        # LRU eviction policy: remove the least recently used item
        self.cache.popitem(last=False)

    def print_statistics(self):
        print(f"Cache Hits: {self.cache_hits}")
        print(f"Cache Misses: {self.cache_misses}")
        print(f"Cache Size: {self.cache_size}")
        # print("Cache Content:")
        # for key, value in self.cache.items():
        #     print(f"{key}: {value}")
        print("\n")

class MRU_Cache:
    def __init__(self, cache_size=10):
        self.cache = OrderedDict()
        self.cache_hits = 0
        self.cache_misses = 0
        self.cache_size = cache_size


    def get(self, key):
        # Works opposite to an LRU where the most recently item used is evicted : Stack behavior 
        if self.cache.get(key, False) != False:
            self.cache_hits += 1
            self.cache.move_to_end(key)  # Move the accessed item to the end
            return self.cache[key]
        else:
            self.cache_misses += 1
            data_from_db = global_database.get(key, False)
            if data_from_db != False:
                if len(self.cache) >= self.cache_size:
                    self.evict()
                self.cache[key] = data_from_db
                return data_from_db

    def evict(self):
        # MRU Eviction Policy: Remove the most recently used item
        self.cache.popitem(last=True)

    def print_statistics(self):
        print(f"Cache Hits: {self.cache_hits}")
        print(f"Cache Misses: {self.cache_misses}")
        print(f"Cache Size: {self.cache_size}")
        # print("Cache Content:")
        # for key, value in self.cache.items():
        #     print(f"{key}: {value}")
        print("\n")  

class FIFO_Cache:
    def __init__(self, cache_size=10):
        self.cache = OrderedDict()
        self.cache_hits = 0
        self.cache_misses = 0
        self.cache_size = cache_size

    def get(self, key):
        if self.cache.get(key, False) != False:
            self.cache_hits += 1
            return self.cache[key]
        else:
            self.cache_misses += 1
            data_from_db = global_database.get(key, False)
            if data_from_db != False:
                if len(self.cache) >= self.cache_size:
                    self.evict()
                self.cache[key] = data_from_db
                return data_from_db

    def evict(self):
        # FIFO eviction policy: remove the first item added to the cache
        self.cache.popitem(last=False)

    def print_statistics(self):
        print(f"Cache Hits: {self.cache_hits}")
        print(f"Cache Misses: {self.cache_misses}")
        print(f"Cache Size: {self.cache_size}")
        # print("Cache Content:")
        # for key, value in self.cache.items():
        #     print(f"{key}: {value}")
        print("\n")

class RR_Cache:
    def __init__(self, cache_size=10):
        self.cache = {}
        self.keys = []
        self.cache_hits = 0
        self.cache_misses = 0
        self.cache_size = cache_size

    def get(self, key):
        if key in self.cache:
            self.cache_hits += 1
            return self.cache[key]
        else:
            self.cache_misses += 1
            data_from_db = global_database.get(key, None)
            if data_from_db is not None:
                if len(self.cache) >= self.cache_size:
                    self.evict()
                self.cache[key] = data_from_db
                self.keys.append(key)
                return data_from_db

    def evict(self):
        remove_idx = randint(0, len(self.keys) - 1)
        remove_key = self.keys.pop(remove_idx)
        del self.cache[remove_key]

    def print_statistics(self):
        print(f"Cache Hits: {self.cache_hits}")
        print(f"Cache Misses: {self.cache_misses}")
        print(f"Cache Size: {self.cache_size}")
        # print("Cache Content:")
        # for item in self.cache:
        #     print(f"{item}")
        print("\n")


```
